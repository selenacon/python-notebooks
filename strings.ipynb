String Fundamentals in Python
==============================
_Code snippets from **Learning Python (5th edition)** by Mark Lutz_

Strings in Python handle ASCII text. Python 3 can also handle Unicode.

Unlike other programing languages, there are no distinctions between strings and characters. Single characters are treated as strings. There also isn't a difference between double and single quotation marks.

Strings are stored as a sequence of characters, and so the operators used to manipulate a string's content can also be applied to other sequences (lists, tuples).

## Common string expressions

Empty string

```python
S = ''
print(S)
```

Double quotes, same as single

```python
S = "spam's"
print(S)
```

Escape sequences

```python
S = 's\np\ta\x00m'
print(S)
```

Triple-quoted block strings

```python
S = """...multiline..."""
print(S)
```

Raw strings (no escapes)

```python
S = r'\temp\spam'
print(S)
```

Concatenate, repeat

```python
S = '!'
S1 = 'Hello, '
S2 = 'world'
print(S1 + S2, S * 3)
```

Index, slice, length

```python
S = 'spam'
print(S[1])
print(S[1:3])
len(S)
```

String formatting expression

```python
kind = "blue"
"a %s parrot" % kind
```

String formatting method in 2.6, 2.7 and 3.X

```python
"a {0} parrot".format(kind)
```

## Single and Double-Quoted Strings Are the Same

```python
'shrubbery', "shribbery"
```

This allows the use of single or double quotes within a string:

```python
'knight"s', "knight's"
```

Most Python programmers use single-quoted strings unless there's the need to embed a single quote in it.

Python concatenates strings by default:

```python
title = "Meaning " 'of' " Life"
title
```

Separating the strings by commas won't concatenate them. Instead, they'll turn into a tuple:

```python
'knight\'s', "knight\"s"
```

_(Note that double and single quotes can be embedded in strings with backslashes as well)_

## Escape Sequences Represent Special Characters

Backslashes are used to introduce escape sequences. These allow the embedding of characters that aren't easily typed with a keyboard.

```python
s = 'a\nb\tc'
print(s)
```

The `len` function returns the number of characters in a string:

```python
len(s)
```
