String Fundamentals in Python
==============================
_Code snippets from **Learning Python (5th edition)** by Mark Lutz_

Strings in Python handle ASCII text. Python 3 can also handle Unicode.

Unlike other programing languages, there are no distinctions between strings and characters. Single characters are treated as strings. There also isn't a difference between double and single quotation marks.

Strings are stored as a sequence of characters, and so the operators used to manipulate a string's content can also be applied to other sequences (lists, tuples).

## Common string expressions

Empty string

```python
S = ''
print(S)
```

Double quotes, same as single

```python
S = "spam's"
print(S)
```

Escape sequences

```python
S = 's\np\ta\x00m'
print(S)
```

Triple-quoted block strings

```python
S = """...multiline..."""
print(S)
```

Raw strings (no escapes)

```python
S = r'\temp\spam'
print(S)
```
