# Mutability
## Immutable Objects
![](img/Fig-6-3.png)

```python
a = 3    # define a
b = a    # assign a's value to b
print('b is', b)

a = 'spam'  #
print('b is', b)
```

a's value has changed, but b's value will stil be 3.

`Int` objects aren't mutable:

```python
a = 3
b = a   # assign a's value to b
a = a + 2   # redefine a as 5
print('b is', b) #
```

b's value doesn't change here either.

## Mutable Objects
### Lists

```python
L = [2, 3, 4]
M = L   # assign L's value to M
L[0] = 24   # redefine the first value in L as 24

print('L =', L)
print('M =', M)
```

`L[0] = 24` changed the value of both variables.

Copying objects instead of referencing them, however, prevents these changes:

```python
L = [2, 3, 4]
M = L[:]   # indexing L makes a copy of it
L[0] = 24

print('L =', L)
print('M =', M)
```

`M` retains its value at the time of the copy.

### Shared References and Equality
- `==` compares values.
- `is` compares object identitites.

```python
L = [1, 2, 3]
M = L

L == M    # does L's value equal to M?
L is M    # do L and M point to the same reference?
```

```python
L = [1, 2 ,3]
M = [1, 2 ,3]

print('L == M: ', L == M)    # True: L and M have the same value
print('L is M: ', L is M)    # False: L and M don't share a reference
```

#### Cache
Small intergers and strings are cashed and reused, therefore, `is` will return `True` even if the variables don't share a reference.

```python
X = 42
Y = 42
X == Y
X is Y
```

`Y` should get its own value, but Python realises `42` is already loaded in memory, so it uses that cached value instead.

Pointers to cached values can be retreived as a reference number with `sys`'s `getrefcount()` function:

```python
import sys
sys.getrefcount(1)
```
