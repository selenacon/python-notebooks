### Pass Mutable Objects

```python
def changer(a, b):
            a = 2
            b[0] = 'spam'    # Change b's first index to 'spam'

x = 1
l = [1, 2]
changer(x, l)
print('l = ', l)
assert x == 1, 'x is 1'    # Check if x = 1
```

### Preventing Rutines from Modifying Mutable Objects
#### Passing a Copy

```python
l = [1, 2]
changer(x, l[:])    # Indexing l makes a copy

print('l = ', l)
```

```python
l = [1, 2]
def changer(a, b):
    b = b[:]
    a = 2
    b[0] = 'spam'

changer(x, l)
assert l == [1,2], 'l es una invariante'
print('l = ', l)
```

```python
l = [1, 2]
def changer(a,b):
        a = 2
        b[0] = 'spam'
changer(x, tuple(l))
assert l == [1, 2], 'l es una invariante'
```

### Special Argument-Matching Modes
#### Keyword and Default Examples
##### 1. Default Behavior

```python
def f(a, b, c):
    print(a, b, c)
f(1, 2, 3)
```

##### 2. Keywords

```python
def f(a, b, c):
    print(a, b, c)
f(c = 3, b = 2, a = 1)
```

```python
f(1, c = 3, b = 2)
```

##### 3. Defaults

```python
def f(a, b = 2, c = 3):
        print(a, b, c)

f(1)
f(a = 1)
```

If we pass two values, only `c` gets its default, and with three values, no defaults are used:

```python
def f(a, b = 2, c = 3):
        print(a, b, c)

f(1, 4)
f(1, 4, 5)

f(1, c = 6)
```

##### 4. Combining Keywords and Defaults

```python
def func(spam, eggs, toast = 0, ham = 0):
        print((spam, eggs, toast, ham))

func(1, 2)
func(1, ham = 1, eggs = 0)
func(spam = 1, eggs = 0)
func(toast = 1, eggs = 2, spam = 3)
func(1, 2, 3, 4)
```

##### 5. Combining Keywords and Defaults

```python
def kwonly(a, *b, c):
    print(a, b, c)
kwonly(1, 2, c = 3)

kwonly(a = 1, c = 3)

kwonly(1, 2, 3)
```
