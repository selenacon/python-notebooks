# Scopes
## Global Scope
Global names are declared outside of any functions and can be used within them.

```python
x = 88


def func():
    x = 99


func()
print(x)
```

`global` allows you to interact with global variables from inside a funcion:

```python
x = 88


def func():
    global x
    x = 99


func()
print("x =", x)
```

## LEGB rule
Global names may be referenced within a function without being declared.

```python
y, z = 1, 2


def all_global():
    global x
    x = y + z


all_global()
print("x = %d, y = %d, z = %d" % (x, y, z))
```

## Nested Scopes

A global and a local variable can have the same name, but they will be treated as different values.

```python
x = 99


def f1():
    x = 88

    def f2():
        print("x local = ", x)

    f2()


f1()
print("x global = ", x)
```

## Factory Functions: Closures
Values will carry over from their enclosing scopes.

```python
def f1():
    x = 88

    def f2():
        print("x =", x)

    return f2


action = f1()
action()
```

```python
def maker(n):
    def action(x):
        return x**n

    return action


f = maker(2)
f(3)
f(4)
```
